# Getting Started with the HSC CIAI Course - Part 1

Welcome to **Catholic Introduction to Artificial Intelligence (HSC CIAI)** course!

All course materials are distributed through the public GitHub repository. This allows you to always have the latest notebooks, tutorials, datasets, and other course resources.

There are two supported ways to use the course materials:

1. **Visual Studio Code (VS Code)** - for past or existing users of VSC; software installations are required. 
2. **Google Colab** - for all others; no software installation is required, only a Google account.

---

## Option 1 – Visual Studio Code

This option is recommended if you want the best experience and plan to complete all course exercises. However, if you have never used VSC and are uncomfortable trying, then use Google Colab.

### Software to Install

Before the first class, install the following software (if you don't already have it installed):

* Python 3.11.9 (there are newer versions of Python; 3.11.9 is recommended for this course)
* Visual Studio Code
* Git

Next, install the VS Code extensions for:

* Python
* Jupyter

---

### Step 1 – Clone the Course Repository

You only need to do this **once**.

Open a terminal (PowerShell or Command Prompt) and navigate to the folder where you want to store your school work.

Example:

```bash
cd Documents
```

Clone the course repository:

```bash
git clone https://github.com/danielgoddu-lang/HSC-CIAI-Course-Part-1.git
```

Git will create a new folder:

```text
Documents
└── HSC-CIAI-Course-Part-1
```

---

### Step 2 – Open the Course in VS Code

Open the repository folder.

Either:

**File → Open Folder**

or

```bash
cd HSC-CIAI-Course-Part-1
code .
```

---

### Step 3 – Install Recommended Extensions

If prompted by VS Code, install the recommended extensions.

Otherwise install:

* Python
* Jupyter

---

### Step 4 – Select the Python Interpreter

The first time you open a notebook:

1. Open a notebook (`.ipynb`)
2. Click **Select Kernel** in the upper-right corner.
3. Choose your installed Python interpreter.

You only need to do this once.

---

### Course Folder Layout

Your repository will look something like this:

```text
HSC-CIAI-Course
│
├── Tutorials
│── Week 1
│── Week 2
│── ...
├── Datasets
└── README.md
```

Each week's lessons will appear in the appropriate folder.

---

### Weekly Updates

This is the biggest advantage of using Git.

**You do NOT download the repository again every week.**

Instead:

Open a terminal inside the course folder and run:

```bash
git pull
```

Git downloads **only the files that have changed**.

If a new lesson is added, it simply appears in your folder after running:

```bash
git pull
```

No ZIP files.

No copying folders.

No replacing files.

---

### If You Made Changes

If you accidentally edit one of the course notebooks, Git may prevent a pull.

If this happens:

* Make a copy of your work.
* Contact the instructor if you are unsure how to proceed.

Students generally should **not** modify the original course notebooks. Instead, save your work with a new filename if instructed.

---

## Option 2 – Google Colab

Google Colab allows you to run Jupyter notebooks entirely in your web browser.

**No software installation is required.**

You only need a Google account.

---

### Opening a Notebook

On the course page in Caravel, click on "Open in new tab -->" for the notebook you want to open.

The notebook will open in a new browser tab (NOTE: you may be asked to log into Google prior to the notebook being opened).

---

## Saving Your Work

Colab notebooks are temporary.

To keep your work:

* Save a copy to your Google Drive, or
* Download the notebook (`.ipynb`) to your computer.

Never edit the instructor's copy directly.

---

### Weekly Updates

Because the notebooks are stored in GitHub, you always open the latest version.

Simply reopen the notebook from GitHub each week.

No software updates are required.

---

### Images and Supporting Files

Many notebooks contain:

* diagrams
* images
* datasets

These files are stored in the GitHub repository alongside the notebook.

When using VS Code, Git automatically downloads these supporting files.

When using Google Colab, notebooks are configured to load these resources directly from the public repository when possible. Some lessons that require local datasets may include additional instructions for downloading or accessing those files.

# Frequently Asked Questions

### Do I need a GitHub account?

**No.**

The course repository is public.

You may clone or download it without creating a GitHub account.

---

### Do I need to download a ZIP file every week?

**No**, if you are using Git.

Simply run:

```bash
git pull
```

Git downloads only the files that have changed.

If you choose not to install Git, you may instead download the latest ZIP from GitHub whenever new materials are posted.

---

### Can I edit the course notebooks?

Yes, but it is recommended that you save your own copy of a notebook before making significant changes. This makes it easy to compare your work with the original course materials and avoids conflicts when new versions are published.

---

# Recommended Workflow

For the best experience throughout the course:

1. Install Python, VS Code, and Git.
2. Clone the course repository once.
3. Open the repository in VS Code.
4. At the beginning of each week, run:

```bash
git pull
```

5. Open the new week's notebook and begin the lesson.

This workflow mirrors how professional software developers and data scientists work, while ensuring you always have the latest course materials with minimal effort.
